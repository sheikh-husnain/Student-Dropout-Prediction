```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("data.csv", sep=";")

df.isnull().sum()

df.duplicated().sum()

df.dtypes

df.select_dtypes(include="object").columns

X = df.drop("Target", axis=1)
X.shape
y = df["Target"]
y.shape

from sklearn.preprocessing import LabelEncoder
label_encoder=LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
categorical_columns = [
    "Marital status",
    "Application mode",
    "Course",
    "Daytime/evening attendance\t",
    "Previous qualification",
    "Nacionality",
    "Mother's qualification",
    "Father's qualification",
    "Mother's occupation",
    "Father's occupation",
    "Displaced",
    "Educational special needs",
    "Debtor",
    "Tuition fees up to date",
    "Gender",
    "Scholarship holder",
    "International"
]
X_encoded = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True
)
X_encoded.isnull().sum().sum()
target_counts = df["Target"].value_counts()
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y_encoded,
    test_size=0.2,
    random_state=42
)
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=10000)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print(y_pred[:10])
print(label_encoder.classes_)
y_prob = model.predict_proba(X_test)
print(y_prob[:5])
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print(accuracy)
from sklearn.metrics import precision_score
precision = precision_score(y_test, y_pred, average="weighted")
print(precision)
from sklearn.metrics import recall_score
recall = recall_score(y_test, y_pred, average="weighted")
print(recall)
from sklearn.metrics import f1_score
f1 = f1_score(y_test, y_pred, average="weighted")
print(f1)
from sklearn.metrics import confusion_matrix, classification_report, roc_auc_score

cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=label_encoder.classes_))

roc_auc = roc_auc_score(
    y_test,
    y_prob,
    multi_class="ovr",
    average="weighted"
)

print("ROC-AUC:", roc_auc)
print(X_encoded.columns.tolist())
student = df.iloc[[0]].drop("Target", axis=1)
student_encoded = pd.get_dummies(
    student,
    columns=categorical_columns,
    drop_first=True
)
student_encoded = student_encoded.reindex(columns=X_encoded.columns, fill_value=0)
print(student_encoded.shape)
prediction = model.predict(student_encoded)
print(prediction)
# Get prediction probabilities
probabilities = model.predict_proba(student_encoded)

# Get the predicted class name
predicted_class = label_encoder.inverse_transform(prediction)[0]

dropout_probability = probabilities[0][0]

if dropout_probability >= 0.70:
    risk_level = "High Risk"
elif dropout_probability >= 0.40:
    risk_level = "Medium Risk"
else:
    risk_level = "Low Risk"
print("Predicted Outcome:", predicted_class)
print("Dropout Probability:", round(dropout_probability * 100, 2), "%")
print("Risk Level:", risk_level)
print("\nAll Class Probabilities:")
print("Dropout:", round(probabilities[0][0] * 100, 2), "%")
print("Enrolled:", round(probabilities[0][1] * 100, 2), "%")
print("Graduate:", round(probabilities[0][2] * 100, 2), "%")
```
